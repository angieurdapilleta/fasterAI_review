## 1-Dataset
This experimentation of FasterAI will be supported by CALTECH-10 dataset. The dataset contains images of people collected from the web by typing common given names into Google Image Search. The coordinates of the eyes, the nose and the center of the mouth for each frontal face are provided in a ground truth file.
We will provide some tools to extract and process the dataset.
Since the dataset provide 304x312 pixels images, there is no need to chip images, raw data will be ok for our model

```python
LOCAL_DATASET_ROOT = Path("")


def load_local_split(split_folder, sequence_name):
    image_location = os.path.join(split_folder ,f"{sequence_name}",)
    label_location = os.path.join(split_folder ,f"{sequence_name}",)
    images, labels = [], []
    for image_doc,label_doc in zip(sorted(os.listdir(image_location)), sorted(os.listdir(label_location))):
        # Full path Doc + File
        image_full_path = os.path.join(image_location, image_doc)
        label_full_path = os.path.join(label_location, label_doc)
        # Images
        images.append(image_full_path)
        # Labels
        labels.append(label_full_path)

    return np.array(images), np.array(labels)

X_train, Y_train = load_local_split(LOCAL_DATASET_ROOT,sequence_name="training")
X_val, Y_val = load_local_split(LOCAL_DATASET_ROOT,sequence_name="validation")
X_test, Y_test = load_local_split(LOCAL_DATASET_ROOT,sequence_name="test")

print("Dataset folder: ", LOCAL_DATASET_ROOT)
print("Training images: ", X_train.shape[0])
print("Validation images: ", X_val.shape[0])
```
